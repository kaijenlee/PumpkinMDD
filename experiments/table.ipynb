```python
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
df_ms = pd.read_csv('width_logs/marketsplit_srv.csv')
df_qcp= pd.read_csv('width_logs/qcp_srv.csv')
```

```python
p = df_ms[["Filename", "reportedMdd", "mddMaxWidth"]].dropna()
p.columns = ['Market Split instance', 'Number of MDDs', 'Maximum MDD width']

print(p.to_latex(
    index=False,  # This removes the index column
    column_format='|c|c|c|',  # Adds borders and centers columns
    escape=False,  # Allows LaTeX commands in data
    caption='Maximum width observed of s3, s4, u3, and u4 instances of Market Split',
    label='tab:MS_widths',
    position='H',  # Requires \usepackage{float}
))
```

```python
p = df_qcp[["Filename", "reportedMdd", "mddMaxWidth"]].dropna()
p.columns = ['QCP instance', 'Number of MDDs', 'Maximum MDD width']

print(p.to_latex(
    index=False,  # This removes the index column
    column_format='|c|c|c|',  # Adds borders and centers columns
    escape=False,  # Allows LaTeX commands in data
    caption='Maximum width observed for QCP instances, with at least 1 MDD successfully constructed.',
    label='tab:_widths',
    position='H',  # Requires \usepackage{float}
))
```
